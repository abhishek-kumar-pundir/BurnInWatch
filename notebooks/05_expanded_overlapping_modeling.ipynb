```python
# ============================================================
# BURNINWATCH — EXPANDED OVERLAPPING-WINDOW EXPERIMENT
# SINGLE MASTER CELL
# ============================================================

from pathlib import Path
import json
import random
import warnings
import joblib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler
from sklearn.ensemble import IsolationForest

import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 220)

# ============================================================
# 1. SETTINGS
# ============================================================

SEED = 42
WINDOW_SIZE = 100
STRIDE = 25

TRAIN_FRAC = 0.20
VAL_FRAC = 0.20

PROCESSED = Path("processed")
MODELS = Path("models")

MODELS.mkdir(exist_ok=True)

RAW_PATH = PROCESSED / "burninwatch_aging_canonical_v1.csv"

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print("=" * 80)
print("BURNINWATCH — EXPANDED MODELING")
print("=" * 80)
print("Device:", DEVICE)
print("Source:", RAW_PATH.resolve())


# ============================================================
# 2. LOAD CANONICAL RAW DATA
# ============================================================

if not RAW_PATH.exists():
    raise FileNotFoundError(
        f"Could not find:\n{RAW_PATH.resolve()}"
    )

raw = pd.read_csv(RAW_PATH)

print("\nRaw shape:", raw.shape)

required = [
    "device_id",
    "file_name",
    "file_role",
    "event_index",
    "timeEpoch",
    "supplyVoltage",
    "node1Voltage",
    "node2Voltage",
    "collectorEmitterCurrent",
    "heatSinkTemperature",
    "packageTemperature",
    "internalTemperature",
    "measurement_valid",
]

missing = [c for c in required if c not in raw.columns]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}"
    )


# ============================================================
# 3. CLEAN RAW DATA
# ============================================================

numeric_cols = [
    "timeEpoch",
    "supplyVoltage",
    "node1Voltage",
    "node2Voltage",
    "collectorEmitterCurrent",
    "heatSinkTemperature",
    "packageTemperature",
    "internalTemperature",
]

for col in numeric_cols:
    raw[col] = pd.to_numeric(
        raw[col],
        errors="coerce"
    )

raw = raw[
    raw["measurement_valid"].astype(bool)
].copy()

raw = raw.dropna(
    subset=numeric_cols
).copy()

raw = raw.sort_values(
    [
        "device_id",
        "file_name",
        "event_index"
    ]
).reset_index(
    drop=True
)

print("Valid raw records:", len(raw))


# ============================================================
# 4. RAW TEMPORAL SPLIT
# ============================================================
#
# SPLIT RAW RECORDS FIRST.
#
# Then windows are created inside each split.
#
# This prevents overlap between train/validation/evaluation.
# ============================================================

raw["split"] = "excluded"

for key, group in raw.groupby(
    ["device_id", "file_name"],
    sort=False
):

    device_id, file_name = key

    group = group.sort_values(
        "event_index"
    )

    indices = group.index.to_numpy()

    n = len(indices)

    role = group[
        "file_role"
    ].iloc[0]

    if role == "check":

        raw.loc[
            indices,
            "split"
        ] = "check_holdout"

        continue

    if n < WINDOW_SIZE * 3:

        raw.loc[
            indices,
            "split"
        ] = "excluded"

        continue

    train_end = int(
        np.floor(
            n * TRAIN_FRAC
        )
    )

    val_end = train_end + int(
        np.floor(
            n * VAL_FRAC
        )
    )

    raw.loc[
        indices[:train_end],
        "split"
    ] = "baseline_train"

    raw.loc[
        indices[train_end:val_end],
        "split"
    ] = "validation"

    raw.loc[
        indices[val_end:],
        "split"
    ] = "evaluation"


print("\nRaw split:")
display(
    raw["split"]
    .value_counts()
    .to_frame("records")
)


# ============================================================
# 5. BASELINE FOR EACH FILE
# ============================================================

def calculate_baseline(group):

    train = group[
        group["split"] == "baseline_train"
    ].copy()

    if len(train) == 0:
        return None

    reference_size = max(
        20,
        int(len(train) * 0.10)
    )

    reference = train.head(
        reference_size
    )

    baseline = {
        "node2Voltage": float(
            reference[
                "node2Voltage"
            ].median()
        ),
        "collectorEmitterCurrent": float(
            reference[
                "collectorEmitterCurrent"
            ].median()
        ),
        "packageTemperature": float(
            reference[
                "packageTemperature"
            ].median()
        ),
    }

    return baseline


baselines = {}

for key, group in raw.groupby(
    ["device_id", "file_name"],
    sort=False
):

    baseline = calculate_baseline(
        group
    )

    if baseline is not None:
        baselines[key] = baseline

print(
    "\nBaselines created:",
    len(baselines)
)


# ============================================================
# 6. FEATURE EXTRACTION
# ============================================================

def calculate_slope(values):

    values = np.asarray(
        values,
        dtype=float
    )

    if len(values) < 2:
        return np.nan

    x = np.arange(
        len(values),
        dtype=float
    )

    x = x - x.mean()

    y = values - values.mean()

    denominator = np.sum(
        x * x
    )

    if denominator == 0:
        return 0.0

    return float(
        np.sum(
            x * y
        )
        / denominator
    )


def extract_features(window, baseline):

    result = {}

    # --------------------------------------------------------
    # BASIC SIGNAL STATISTICS
    # --------------------------------------------------------

    signals = [
        "supplyVoltage",
        "node2Voltage",
        "collectorEmitterCurrent",
        "packageTemperature",
    ]

    for signal in signals:

        values = window[
            signal
        ].to_numpy(
            dtype=float
        )

        result[
            signal + "_mean"
        ] = float(
            values.mean()
        )

        result[
            signal + "_std"
        ] = float(
            values.std()
        )

        result[
            signal + "_q05"
        ] = float(
            np.quantile(
                values,
                0.05
            )
        )

        result[
            signal + "_q95"
        ] = float(
            np.quantile(
                values,
                0.95
            )
        )

        result[
            signal + "_iqr"
        ] = float(
            np.quantile(
                values,
                0.75
            )
            -
            np.quantile(
                values,
                0.25
            )
        )

        result[
            signal + "_slope"
        ] = calculate_slope(
            values
        )

    # --------------------------------------------------------
    # RELATIONSHIP FEATURES
    # --------------------------------------------------------

    supply_mean = result[
        "supplyVoltage_mean"
    ]

    node2_mean = result[
        "node2Voltage_mean"
    ]

    current_mean = result[
        "collectorEmitterCurrent_mean"
    ]

    node1_mean = float(
        window[
            "node1Voltage"
        ].mean()
    )

    result[
        "supply_node1_mean_diff"
    ] = supply_mean - node1_mean

    result[
        "node1_node2_mean_diff"
    ] = node1_mean - node2_mean

    result[
        "node_voltage_drop"
    ] = node1_mean - node2_mean

    result[
        "current_per_node2V"
    ] = (
        current_mean
        /
        max(
            abs(node2_mean),
            1e-9
        )
    )

    result[
        "estimated_power_proxy"
    ] = (
        node2_mean
        *
        current_mean
    )

    # --------------------------------------------------------
    # CHANGE FROM EARLY BASELINE
    # --------------------------------------------------------

    for signal in [
        "node2Voltage",
        "collectorEmitterCurrent",
        "packageTemperature"
    ]:

        current_value = result[
            signal + "_mean"
        ]

        reference = baseline[
            signal
        ]

        delta = (
            current_value
            -
            reference
        )

        percentage = (
            delta
            /
            max(
                abs(reference),
                1e-9
            )
        )

        result[
            signal + "_delta_from_baseline"
        ] = delta

        result[
            signal + "_pct_change_from_baseline"
        ] = percentage

    return result


# ============================================================
# 7. CREATE OVERLAPPING WINDOWS
# ============================================================

window_rows = []

for key, group in raw.groupby(
    ["device_id", "file_name"],
    sort=False
):

    device_id, file_name = key

    group = group.sort_values(
        "event_index"
    )

    baseline = baselines.get(
        key
    )

    if baseline is None:
        continue

    for split_name in [
        "baseline_train",
        "validation",
        "evaluation",
        "check_holdout"
    ]:

        segment = group[
            group["split"] == split_name
        ].copy()

        segment = segment.reset_index(
            drop=True
        )

        if len(segment) < WINDOW_SIZE:
            continue

        local_window_index = 0

        max_start = (
            len(segment)
            -
            WINDOW_SIZE
            +
            1
        )

        for start in range(
            0,
            max_start,
            STRIDE
        ):

            end = (
                start
                +
                WINDOW_SIZE
            )

            window = segment.iloc[
                start:end
            ]

            features = extract_features(
                window,
                baseline
            )

            row = {
                "device_id": device_id,
                "file_name": file_name,
                "file_role": window[
                    "file_role"
                ].iloc[0],
                "split": split_name,
                "window_index": local_window_index,
                "window_start_event": int(
                    window[
                        "event_index"
                    ].iloc[0]
                ),
                "window_end_event": int(
                    window[
                        "event_index"
                    ].iloc[-1]
                ),
                "window_size": len(window),
                "start_time": float(
                    window[
                        "timeEpoch"
                    ].iloc[0]
                ),
                "end_time": float(
                    window[
                        "timeEpoch"
                    ].iloc[-1]
                ),
            }

            row.update(
                features
            )

            window_rows.append(
                row
            )

            local_window_index += 1


windows = pd.DataFrame(
    window_rows
)

print(
    "\nExpanded windows:",
    windows.shape
)

print("\nWindows by split:")
display(
    windows["split"]
    .value_counts()
    .to_frame("windows")
)

print("\nWindows by device:")
display(
    windows.groupby(
        ["device_id", "split"]
    )
    .size()
    .unstack(
        fill_value=0
    )
)


# ============================================================
# 8. DEFINE MODEL FEATURES
# ============================================================

META = [
    "device_id",
    "file_name",
    "file_role",
    "split",
    "window_index",
    "window_start_event",
    "window_end_event",
    "window_size",
    "start_time",
    "end_time",
]

EXPANDED_FEATURES = [
    "supplyVoltage_mean",
    "supplyVoltage_std",

    "node2Voltage_mean",
    "node2Voltage_std",
    "node2Voltage_q05",
    "node2Voltage_q95",
    "node2Voltage_iqr",
    "node2Voltage_slope",

    "collectorEmitterCurrent_mean",
    "collectorEmitterCurrent_std",
    "collectorEmitterCurrent_q05",
    "collectorEmitterCurrent_q95",
    "collectorEmitterCurrent_iqr",
    "collectorEmitterCurrent_slope",

    "packageTemperature_mean",
    "packageTemperature_std",
    "packageTemperature_q05",
    "packageTemperature_q95",
    "packageTemperature_iqr",
    "packageTemperature_slope",

    "supply_node1_mean_diff",
    "node1_node2_mean_diff",
    "node_voltage_drop",
    "current_per_node2V",
    "estimated_power_proxy",

    "node2Voltage_delta_from_baseline",
    "node2Voltage_pct_change_from_baseline",

    "collectorEmitterCurrent_delta_from_baseline",
    "collectorEmitterCurrent_pct_change_from_baseline",

    "packageTemperature_delta_from_baseline",
    "packageTemperature_pct_change_from_baseline",
]

RESPONSE_FEATURES = [
    "node2Voltage_std",
    "node2Voltage_iqr",
    "node2Voltage_slope",

    "collectorEmitterCurrent_mean",
    "collectorEmitterCurrent_std",
    "collectorEmitterCurrent_q05",
    "collectorEmitterCurrent_q95",
    "collectorEmitterCurrent_iqr",
    "collectorEmitterCurrent_slope",

    "packageTemperature_std",
    "packageTemperature_iqr",
    "packageTemperature_slope",

    "node2Voltage_delta_from_baseline",
    "node2Voltage_pct_change_from_baseline",

    "collectorEmitterCurrent_delta_from_baseline",
    "collectorEmitterCurrent_pct_change_from_baseline",

    "packageTemperature_delta_from_baseline",
    "packageTemperature_pct_change_from_baseline",

    "current_per_node2V",
    "node_voltage_drop",
]

FEATURE_SETS = {
    "expanded": EXPANDED_FEATURES,
    "response_focused": RESPONSE_FEATURES,
}

for name, feature_list in FEATURE_SETS.items():

    missing_features = [
        c
        for c in feature_list
        if c not in windows.columns
    ]

    if missing_features:
        raise ValueError(
            f"{name} missing: {missing_features}"
        )

    print(
        f"{name}: {len(feature_list)} features"
    )


# ============================================================
# 9. PYTORCH AUTOENCODER
# ============================================================

class BurnInAutoencoder(nn.Module):

    def __init__(self, input_dim):

        super().__init__()

        hidden_dim = max(
            16,
            min(
                64,
                input_dim * 2
            )
        )

        bottleneck_dim = max(
            4,
            min(
                12,
                input_dim // 3
            )
        )

        self.encoder = nn.Sequential(
            nn.Linear(
                input_dim,
                hidden_dim
            ),
            nn.ReLU(),

            nn.Linear(
                hidden_dim,
                bottleneck_dim
            )
        )

        self.decoder = nn.Sequential(
            nn.Linear(
                bottleneck_dim,
                hidden_dim
            ),
            nn.ReLU(),

            nn.Linear(
                hidden_dim,
                input_dim
            )
        )

    def forward(self, x):

        encoded = self.encoder(
            x
        )

        reconstructed = self.decoder(
            encoded
        )

        return reconstructed


def train_autoencoder(
    X,
    input_dim,
    epochs=150
):

    dataset = TensorDataset(
        torch.tensor(
            X,
            dtype=torch.float32
        )
    )

    loader = DataLoader(
        dataset,
        batch_size=min(
            64,
            len(dataset)
        ),
        shuffle=True
    )

    model = BurnInAutoencoder(
        input_dim
    ).to(DEVICE)

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=1e-3,
        weight_decay=1e-4
    )

    criterion = nn.MSELoss()

    losses = []

    for epoch in range(
        epochs
    ):

        model.train()

        running_loss = 0.0

        for (
            batch,
        ) in loader:

            batch = batch.to(
                DEVICE
            )

            optimizer.zero_grad()

            reconstruction = model(
                batch
            )

            loss = criterion(
                reconstruction,
                batch
            )

            loss.backward()

            optimizer.step()

            running_loss += (
                loss.item()
                *
                len(batch)
            )

        epoch_loss = (
            running_loss
            /
            len(dataset)
        )

        losses.append(
            epoch_loss
        )

    return model, losses


@torch.no_grad()
def autoencoder_score(
    model,
    X
):

    model.eval()

    tensor = torch.tensor(
        X,
        dtype=torch.float32
    ).to(DEVICE)

    reconstructed = model(
        tensor
    )

    error = (
        tensor
        -
        reconstructed
    ) ** 2

    return (
        error
        .mean(
            dim=1
        )
        .cpu()
        .numpy()
    )


# ============================================================
# 10. RUN MODELS
# ============================================================

comparison_rows = []
losses_by_experiment = {}

for experiment_name, feature_list in FEATURE_SETS.items():

    print(
        "\n" + "=" * 80
    )

    print(
        "RUNNING:",
        experiment_name
    )

    print(
        "=" * 80
    )

    train_mask = (
        windows["split"]
        == "baseline_train"
    )

    validation_mask = (
        windows["split"]
        == "validation"
    )

    evaluation_mask = (
        windows["split"]
        == "evaluation"
    )

    check_mask = (
        windows["split"]
        == "check_holdout"
    )

    X_train_raw = windows.loc[
        train_mask,
        feature_list
    ].to_numpy(
        dtype=float
    )

    X_validation_raw = windows.loc[
        validation_mask,
        feature_list
    ].to_numpy(
        dtype=float
    )

    X_evaluation_raw = windows.loc[
        evaluation_mask,
        feature_list
    ].to_numpy(
        dtype=float
    )

    # --------------------------------------------------------
    # SHARED PREPROCESSING
    # --------------------------------------------------------

    imputer = SimpleImputer(
        strategy="median"
    )

    scaler = RobustScaler()

    X_train = imputer.fit_transform(
        X_train_raw
    )

    X_train = scaler.fit_transform(
        X_train
    )

    X_validation = scaler.transform(
        imputer.transform(
            X_validation_raw
        )
    )

    X_evaluation = scaler.transform(
        imputer.transform(
            X_evaluation_raw
        )
    )

    # --------------------------------------------------------
    # ISOLATION FOREST
    # --------------------------------------------------------

    isolation_forest = IsolationForest(
        n_estimators=500,
        contamination="auto",
        random_state=SEED,
        n_jobs=-1
    )

    isolation_forest.fit(
        X_train
    )

    if_validation_scores = (
        -isolation_forest.decision_function(
            X_validation
        )
    )

    if_evaluation_scores = (
        -isolation_forest.decision_function(
            X_evaluation
        )
    )

    if_threshold = float(
        np.quantile(
            if_validation_scores,
            0.95
        )
    )

    if_flags = (
        if_evaluation_scores
        >
        if_threshold
    )

    # --------------------------------------------------------
    # AUTOENCODER
    # --------------------------------------------------------

    autoencoder, losses = train_autoencoder(
        X_train,
        len(feature_list),
        epochs=150
    )

    losses_by_experiment[
        experiment_name
    ] = losses

    ae_validation_scores = autoencoder_score(
        autoencoder,
        X_validation
    )

    ae_evaluation_scores = autoencoder_score(
        autoencoder,
        X_evaluation
    )

    ae_threshold = float(
        np.quantile(
            ae_validation_scores,
            0.95
        )
    )

    ae_flags = (
        ae_evaluation_scores
        >
        ae_threshold
    )

    # --------------------------------------------------------
    # STORE SCORES
    # --------------------------------------------------------

    if_score_column = (
        experiment_name
        +
        "_IF_score"
    )

    ae_score_column = (
        experiment_name
        +
        "_AE_score"
    )

    if_flag_column = (
        experiment_name
        +
        "_IF_flag"
    )

    ae_flag_column = (
        experiment_name
        +
        "_AE_flag"
    )

    windows[if_score_column] = np.nan
    windows[ae_score_column] = np.nan

    windows.loc[
        validation_mask,
        if_score_column
    ] = if_validation_scores

    windows.loc[
        evaluation_mask,
        if_score_column
    ] = if_evaluation_scores

    windows.loc[
        validation_mask,
        ae_score_column
    ] = ae_validation_scores

    windows.loc[
        evaluation_mask,
        ae_score_column
    ] = ae_evaluation_scores

    windows[if_flag_column] = (
        windows[
            if_score_column
        ]
        >
        if_threshold
    ).astype("Int64")

    windows[ae_flag_column] = (
        windows[
            ae_score_column
        ]
        >
        ae_threshold
    ).astype("Int64")

    # --------------------------------------------------------
    # MODEL AGREEMENT
    # --------------------------------------------------------

    both = int(
        (
            if_flags
            &
            ae_flags
        ).sum()
    )

    if_only = int(
        (
            if_flags
            &
            ~ae_flags
        ).sum()
    )

    ae_only = int(
        (
            ~if_flags
            &
            ae_flags
        ).sum()
    )

    correlation = float(
        pd.Series(
            if_evaluation_scores
        ).corr(
            pd.Series(
                ae_evaluation_scores
            )
        )
    )

    comparison_rows.append({

        "experiment":
            experiment_name,

        "feature_count":
            len(feature_list),

        "train_windows":
            len(X_train),

        "validation_windows":
            len(X_validation),

        "evaluation_windows":
            len(X_evaluation),

        "IF_threshold":
            if_threshold,

        "IF_flagged":
            int(if_flags.sum()),

        "IF_flag_fraction":
            float(if_flags.mean()),

        "AE_threshold":
            ae_threshold,

        "AE_flagged":
            int(ae_flags.sum()),

        "AE_flag_fraction":
            float(ae_flags.mean()),

        "both_models_flagged":
            both,

        "IF_only":
            if_only,

        "AE_only":
            ae_only,

        "score_correlation":
            correlation,

        "AE_final_loss":
            float(losses[-1])
    })

    # --------------------------------------------------------
    # SAVE MODEL BUNDLES
    # --------------------------------------------------------

    joblib.dump(
        {
            "imputer": imputer,
            "scaler": scaler,
            "model": isolation_forest,
            "features": feature_list,
            "threshold": if_threshold,
        },
        MODELS
        /
        f"burninwatch_expanded_{experiment_name}_IF.joblib"
    )

    torch.save(
        {
            "state_dict":
                autoencoder.state_dict(),

            "input_dim":
                len(feature_list),

            "features":
                feature_list,

            "threshold":
                ae_threshold,
        },
        MODELS
        /
        f"burninwatch_expanded_{experiment_name}_AE.pt"
    )

    joblib.dump(
        {
            "imputer":
                imputer,

            "scaler":
                scaler,

            "features":
                feature_list,
        },
        MODELS
        /
        f"burninwatch_expanded_{experiment_name}_AE_preprocess.joblib"
    )


# ============================================================
# 11. FINAL COMPARISON
# ============================================================

comparison = pd.DataFrame(
    comparison_rows
)

print(
    "\n" + "=" * 80
)

print(
    "FINAL MODEL COMPARISON"
)

print(
    "=" * 80
)

display(
    comparison
)


# ============================================================
# 12. AUTOENCODER LOSS CURVES
# ============================================================

for experiment_name, losses in losses_by_experiment.items():

    plt.figure(
        figsize=(9, 4)
    )

    plt.plot(
        losses
    )

    plt.title(
        experiment_name
        +
        " — Autoencoder loss"
    )

    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        "Reconstruction MSE"
    )

    plt.grid(
        alpha=0.25
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 13. ANOMALY TIMELINES
# ============================================================

for experiment_name in FEATURE_SETS.keys():

    for device_id, group in windows[
        windows["split"] == "evaluation"
    ].groupby(
        "device_id"
    ):

        group = group.sort_values(
            [
                "file_name",
                "window_start_event"
            ]
        ).reset_index(
            drop=True
        )

        if len(group) == 0:
            continue

        x = np.arange(
            len(group)
        )

        plt.figure(
            figsize=(14, 4)
        )

        plt.plot(
            x,
            group[
                experiment_name
                +
                "_IF_score"
            ],
            label="Isolation Forest",
            linewidth=0.8
        )

        plt.plot(
            x,
            group[
                experiment_name
                +
                "_AE_score"
            ],
            label="Autoencoder",
            linewidth=0.8
        )

        plt.title(
            experiment_name
            +
            " — "
            +
            str(device_id)
        )

        plt.xlabel(
            "Chronological evaluation window"
        )

        plt.ylabel(
            "Model score"
        )

        plt.legend()

        plt.grid(
            alpha=0.25
        )

        plt.tight_layout()

        plt.show()


# ============================================================
# 14. TOP ANOMALIES
# ============================================================

for experiment_name in FEATURE_SETS.keys():

    print(
        "\n" + "=" * 80
    )

    print(
        "TOP ISOLATION FOREST ANOMALIES:",
        experiment_name
    )

    print(
        "=" * 80
    )

    if_score_column = (
        experiment_name
        +
        "_IF_score"
    )

    if_flag_column = (
        experiment_name
        +
        "_IF_flag"
    )

    top_if = (
        windows[
            windows["split"] == "evaluation"
        ]
        .sort_values(
            if_score_column,
            ascending=False
        )
        .head(15)
    )

    display(
        top_if[
            META
            +
            [
                if_score_column,
                if_flag_column,
            ]
        ]
    )

    print(
        "TOP AUTOENCODER ANOMALIES:",
        experiment_name
    )

    ae_score_column = (
        experiment_name
        +
        "_AE_score"
    )

    ae_flag_column = (
        experiment_name
        +
        "_AE_flag"
    )

    top_ae = (
        windows[
            windows["split"] == "evaluation"
        ]
        .sort_values(
            ae_score_column,
            ascending=False
        )
        .head(15)
    )

    display(
        top_ae[
            META
            +
            [
                ae_score_column,
                ae_flag_column,
            ]
        ]
    )


# ============================================================
# 15. SAVE RESULTS
# ============================================================

expanded_output = (
    PROCESSED
    /
    "burninwatch_expanded_overlapping_windows_v2.csv"
)

comparison_output = (
    PROCESSED
    /
    "burninwatch_expanded_model_comparison_v2.csv"
)

windows.to_csv(
    expanded_output,
    index=False
)

comparison.to_csv(
    comparison_output,
    index=False
)

artifact_json = (
    MODELS
    /
    "burninwatch_expanded_artifacts_v2.json"
)

with open(
    artifact_json,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        {
            "window_size":
                WINDOW_SIZE,

            "stride":
                STRIDE,

            "train_fraction":
                TRAIN_FRAC,

            "validation_fraction":
                VAL_FRAC,

            "feature_sets":
                FEATURE_SETS,

            "device":
                DEVICE,

            "comparison":
                comparison_rows,
        },
        f,
        indent=2
    )


# ============================================================
# 16. FINAL CHECK
# ============================================================

print(
    "\n" + "=" * 80
)

print(
    "✅ BURNINWATCH EXPANDED EXPERIMENT COMPLETE"
)

print(
    "=" * 80
)

print(
    "Raw records:",
    len(raw)
)

print(
    "Expanded windows:",
    len(windows)
)

print(
    "Training windows:",
    int(
        (
            windows["split"]
            ==
            "baseline_train"
        ).sum()
    )
)

print(
    "Validation windows:",
    int(
        (
            windows["split"]
            ==
            "validation"
        ).sum()
    )
)

print(
    "Evaluation windows:",
    int(
        (
            windows["split"]
            ==
            "evaluation"
        ).sum()
    )
)

print(
    "\nSaved:"
)

print(
    expanded_output
)

print(
    comparison_output
)

print(
    artifact_json
)

print(
    "\n✅ STOP HERE — SEND ME THE FINAL MODEL COMPARISON TABLE."
)
```
